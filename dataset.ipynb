```python
import json
from tqdm import tqdm

def apply_template_mistral_instruct(system_message, content):
    prompt = f"{system_message}\n{content}".strip()
    return f"[INST] {prompt} [/INST]"

data = []
with open("/mnt/nas/suehyun/MPA/data/train/preferences_v1_responses_for_orpo_64k_v2_ppo.jsonl") as f:
    for line in tqdm(f):
        d = json.loads(line)
        data.append({
            "input": apply_template_mistral_instruct(d["system"], d["instruction"]),
        })
        
with open("/mnt/nas/suehyun/MPA/data/train/preferences_v1_responses_for_orpo_64k_v2_ppo_mistral_input.jsonl", 'w') as f:
    for d in tqdm(data):
        f.write(json.dumps(d) + '\n')
```
